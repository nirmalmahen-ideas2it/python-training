```python
for i in range(1,11):
    if i%2==0:
        print("Number: "+str(i)+" is even")
    elif i%2!=0:
        print("Number: "+str(i)+" is odd")
```

```python
oddList=[x for x in range(1,21) if x%3==0 and x%2!=0]
print(oddList)

nums = [1, 2, 2, 3, 3, 4, 4, 5]
evenSet={x*x for x in nums if x%2==0}
print(evenSet)
```

```python
lists=[1, 2, 2, 3, 3, 4, 4, 5]
sets={x*x for x in lists if x%2==0}
print(sets)
```

```python
dicts={i:i**3 for i in range(1,6)}
print(dicts)
```

```python
def fun(x):
    if(x==0):
        return 0
    elif(x==1):
        return 1
    else:
        return fun(x-1)+fun(x-2)
    
print(fun(8))
    
```

```python
words = ['cat', 'apple', 'banana', 'dog']

sorted_words = sorted(words, key=lambda word: word[-1])
print(sorted_words)  # ['banana', 'apple', 'cat', 'dog']
```
