```python
def generate_report(*subject, **marks):
    for sub in subject:
        print(f"Subject: {sub}   Marks: {marks[sub]}")
generate_report("Maths", "English", Maths=99, English=89, Science=92)
```

```python
def counter():
    count = 0
    def increment():
        nonlocal count
        count += 1  # Error: UnboundLocalError
        return count
    return increment
c = counter()
b=counter()
print(counter().__closure__[0].cell_contents)
print(counter())
print(c())
print(c())
print(b())
```

```python
#Write up Closure function.
# A Closure function is a function which pratically has memory of its instance
#  and its values:

def counter():
    count=0
    def increment():
        nonlocal count
        count+=1
        return count
    return increment

c=counter()
print (c())
print(f"Closure Var: {c.__closure__[0].cell_contents}")
print (c())
print(f"Closure Var: {c.__closure__[0].cell_contents}")

print("================================================================")

def power(p):
    def calc(x):
        return x**p
    return calc

square=power(2)
print(f"Sqaure: {square(5)}")
```
